```python
import pickle
from pathlib import Path

# 1. 기존 2004 R1 -> 2006으로 remapping됐던 56개
old_path = (
    ROOT / "runs"
    / "predecessor_2004_to_2006"
    / "source_round_01"
    / "inherited.pkl"
)

with open(old_path, "rb") as f:
    old = pickle.load(f)

old_pairs = {
    tuple(map(str, p))
    for p in old["inherited_patent_pairs"]
}

# 2. checkpoint-only 2006 R0에서 새로 선택된 pseudo
new_path = (
    ROOT / "runs_checkpoint_only"
    / "cutoff_2006"
    / "round_00"
    / "pseudo_selected_for_round_01.pkl"
)

with open(new_path, "rb") as f:
    new = pickle.load(f)

print("NEW TYPE:", type(new))
print("NEW KEYS:", new.keys() if isinstance(new, dict) else "not dict")

print("\n기존 remapped pseudo:", len(old_pairs))
```

```python
print("pseudo_mp:", len(new["pseudo_mp"]))
print("pseudo_sup:", len(new["pseudo_sup"]))

if new["pseudo_mp"]:
    print("\n[MP sample]")
    print(new["pseudo_mp"][0])

if new["pseudo_sup"]:
    print("\n[SUP sample]")
    print(new["pseudo_sup"][0])
```

```python
# 새로 선택된 34개 pseudo의 patent pair
new_records = new["pseudo_mp"] + new["pseudo_sup"]

new_pairs = {
    (str(r["plt_patent"]), str(r["def_patent"]))
    for r in new_records
}

# 기존 56개와 교집합
recovered = old_pairs & new_pairs
not_recovered = old_pairs - new_pairs
new_only = new_pairs - old_pairs

print("=" * 70)
print("CHECKPOINT-ONLY PSEUDO RECOVERY")
print("=" * 70)

print(f"기존 remapped pseudo       : {len(old_pairs)}")
print(f"새로 선택된 pseudo         : {len(new_pairs)}")
print(f"기존 pseudo 재선택         : {len(recovered)}")
print(f"기존 pseudo 미재선택       : {len(not_recovered)}")
print(f"새롭게 선택된 다른 pseudo  : {len(new_only)}")

recovery_rate = len(recovered) / len(old_pairs) * 100

print(f"\nRecovery Rate             : {recovery_rate:.2f}%")

print("\n[재선택된 기존 pseudo]")
for p in sorted(recovered):
    print(p)
```
